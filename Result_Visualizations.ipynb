```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from scipy.spatial.distance import pdist

# Data for hierarchical heatmap
data = {
    'Model': ['IndicBERT', 'XLM-RoBERTa', 'MuRIL', 'M-BERT', 'HiENBERT-Indic', 'HiENBERT-XLMR'],
    'Accuracy': [0.9144, 0.9321, 0.9351, 0.9308, 0.9167, 0.9328],
    'F1-Macro': [0.9147, 0.9334, 0.9365, 0.9325, 0.9166, 0.9342],
    'Precision': [0.9152, 0.9330, 0.9355, 0.9315, 0.9171, 0.9332],
    'Recall': [0.9144, 0.9321, 0.9351, 0.9308, 0.9167, 0.9328],
    "Cohen's Kappa": [0.9010, 0.9216, 0.9251, 0.9201, 0.9037, 0.9224],
    'Mathews CC': [0.9012, 0.9217, 0.9252, 0.9202, 0.9038, 0.9225]
}

df = pd.DataFrame(data)
df.set_index('Model', inplace=True)

# Transpose for better visualization
df_t = df.T

# Create figure with dendrogram
fig = plt.figure(figsize=(14, 10))

# Compute linkage for row clustering (metrics)
row_linkage = linkage(pdist(df_t.values), method='average')

# Compute linkage for column clustering (models)
col_linkage = linkage(pdist(df_t.T.values), method='average')

# Create heatmap with hierarchical clustering
g = sns.clustermap(df_t, 
                   row_cluster=True, col_cluster=True,
                   row_linkage=row_linkage, col_linkage=col_linkage,
                   annot=True, fmt='.3f', cmap='RdYlGn',
                   linewidths=0.5, linecolor='white',
                   cbar_kws={'label': 'Score', 'shrink': 0.7},
                   vmin=0.88, vmax=0.95,
                   annot_kws={'size': 10},
                   figsize=(12, 8),
                   dendrogram_ratio=0.2)

g.ax_heatmap.set_xlabel('Model', fontsize=14, fontweight='bold')
g.ax_heatmap.set_ylabel('Metric', fontsize=14, fontweight='bold')
g.ax_heatmap.set_xticklabels(g.ax_heatmap.get_xticklabels(), rotation=15, ha='right', fontsize=11)
g.ax_heatmap.set_yticklabels(g.ax_heatmap.get_yticklabels(), fontsize=11)

plt.tight_layout()
plt.savefig('figure3_hierarchical_heatmap.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
models = ['IndicBERT', 'XLM-RoBERTa', 'MuRIL', 'M-BERT', 'HiENBERT-Indic', 'HiENBERT-XLMR']
original_f1 = [0.7580, 0.7735, 0.7598, 0.7590, 0.7241, 0.7276]
combined_f1 = [0.9147, 0.9334, 0.9365, 0.9325, 0.9166, 0.9342]

# Calculate improvement percentage
improvement = [((c - o) / o) * 100 for c, o in zip(combined_f1, original_f1)]

# Create figure
fig, ax = plt.subplots(figsize=(12, 7))

# Create horizontal bar chart with gradient colors
colors = plt.cm.Greens(np.linspace(0.4, 0.9, len(models)))
bars = ax.barh(models, improvement, color=colors, edgecolor='black', linewidth=1.5, height=0.6)

# Add value labels
for i, (bar, value) in enumerate(zip(bars, improvement)):
    ax.text(bar.get_width() + 0.5, bar.get_y() + bar.get_height()/2, 
            f'{value:.1f}%', va='center', ha='left', fontsize=12, fontweight='bold')
    
    # Add small arrow to show improvement trend
    if value > 0:
        ax.annotate('▲', xy=(bar.get_width() - 2, bar.get_y() + bar.get_height()/2), 
                    ha='right', va='center', fontsize=14, color='darkgreen')

# Add baseline line
ax.axvline(x=0, color='black', linestyle='-', linewidth=1.5)

ax.set_xlabel('Relative Improvement (%)', fontsize=14, fontweight='bold')
ax.set_ylabel('Model', fontsize=14, fontweight='bold')
ax.set_xlim(-5, max(improvement) + 5)
ax.grid(True, alpha=0.3, linestyle='--', axis='x')
ax.set_title('', fontsize=16, fontweight='bold')

# Add annotation for highest improvement
max_idx = np.argmax(improvement)
max_model = models[max_idx]
max_imp = improvement[max_idx]
ax.annotate(f'Best: {max_model}', 
            xy=(max_imp, max_idx), xytext=(max_imp + 8, max_idx + 0.2),
            fontsize=11, fontweight='bold', color='darkgreen',
            arrowprops=dict(arrowstyle='->', color='darkgreen', lw=1.5))

plt.tight_layout()
plt.savefig('figure4_relative_improvement_barchart.png', dpi=300, bbox_inches='tight')
plt.show()
```
